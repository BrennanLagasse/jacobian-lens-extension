```python
import torch
```

```python
def remove_projection(H, x):
    """
    H: (v, n)
    x: (n,)

    Returns:
        H_perp: (v, n), with the component along x removed
    """
    return H - ((H @ x) / torch.dot(x, x)).unsqueeze(1) * x


v = 5
n = 2

H = torch.rand((v, n))
x = torch.tensor([1.0, 0.0])

out = remove_projection(H, x)

print(H)
print(x)
print(out)
```
